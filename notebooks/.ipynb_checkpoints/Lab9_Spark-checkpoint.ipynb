```python
import pandas as pd, unicodedata, re
from pyspark.sql import functions as F
from pyspark.sql import SparkSession

# Spark (si no existe)
spark = SparkSession.builder.appName("lab9").getOrCreate()

def norm_cols(cols):
    out = []
    for c in cols:
        c = "".join(ch for ch in unicodedata.normalize("NFD", str(c)) if unicodedata.category(ch) != "Mn")
        c = c.strip().lower().replace("-", "_").replace(" ", "_")
        c = re.sub(r"[^a-z0-9_]", "", c)
        out.append(c)
    return out

def read_excel_to_spark_safe(path):
    """
    Lee Excel con pandas dtype=str para evitar tipos mixtos
    y convierte a Spark. Luego castearás columnas en Spark.
    """
    pdf = pd.read_excel(path, engine="openpyxl", dtype=str)
    pdf.columns = norm_cols(pdf.columns)
    # Normaliza vacíos: '' o 'nan' -> None (para poder castear en Spark)
    pdf = pdf.applymap(lambda x: None if (x is None or str(x).strip() in {"", "nan", "None"}) else str(x).strip())
    return spark.createDataFrame(pdf)

def cast_if(df, col, dtype):
    return df.withColumn(col, F.col(col).cast(dtype)) if col in df.columns else df

def cast_if_many(df, casts: dict):
    """
    casts = {"hora":"int", "num_vehiculos":"int", "lesionados":"int", "fallecidos":"int", "anio":"int"}
    """
    for c, t in casts.items():
        if c in df.columns:
            df = df.withColumn(c, F.col(c).cast(t))
    return df
```

```python
BASE = "/opt/app/working_dir/"
path_grand = f"{BASE}/grand_dataframe.xlsx"

df_grand = read_excel_to_spark_safe(path_grand)

# Casteos suaves
df_grand = cast_if(df_grand, "anio", "int")
df_grand = cast_if(df_grand, "dia_ocu", "int")
df_grand = cast_if(df_grand, "zona_ocu", "int")
df_grand = cast_if(df_grand, "num_vehiculos_involucrados", "int")
# hora_ocu la dejamos como string

df_grand.createOrReplaceTempView("grand")

df_grand.printSchema()
df_grand.show(10, truncate=False)
```

```python
from pyspark.sql import functions as F

# mapa meses (español -> número)
mes_map = F.create_map(
    F.lit("enero"), F.lit(1), F.lit("febrero"), F.lit(2), F.lit("marzo"), F.lit(3),
    F.lit("abril"), F.lit(4), F.lit("mayo"), F.lit(5), F.lit("junio"), F.lit(6),
    F.lit("julio"), F.lit(7), F.lit("agosto"), F.lit(8), F.lit("septiembre"), F.lit(9),
    F.lit("octubre"), F.lit(10), F.lit("noviembre"), F.lit(11), F.lit("diciembre"), F.lit(12)
)

g = df_grand

# Deriva fecha y día de semana
g = (g
     .withColumn("mes_l", F.lower(F.col("mes_ocu")))
     .withColumn("mes_num", mes_map[F.col("mes_l")].cast("int"))
     .withColumn("fecha_ocu",
                 F.to_date(F.format_string("%04d-%02d-%02d",
                                           F.col("anio"), F.col("mes_num"), F.col("dia_ocu"))))
     .withColumn("dow_en", F.date_format(F.col("fecha_ocu"), "E"))
     .withColumn(
         "dia_semana",
         F.when(F.col("dow_en").isin("Mon","Mon."),"Lunes")
          .when(F.col("dow_en").isin("Tue","Tue."),"Martes")
          .when(F.col("dow_en").isin("Wed","Wed."),"Miércoles")
          .when(F.col("dow_en").isin("Thu","Thu."),"Jueves")
          .when(F.col("dow_en").isin("Fri","Fri."),"Viernes")
          .when(F.col("dow_en").isin("Sat","Sat."),"Sábado")
          .when(F.col("dow_en").isin("Sun","Sun."),"Domingo")
     )
)

# Selección final para el análisis exploratorio
select_cols = [
    "anio","mes_ocu","dia_ocu","dia_semana","hora_ocu","zona_ocu",
    "depto_ocu","tipo_eve","num_vehiculos_involucrados","fall_les","origen"
]
df_preparado = g.select(*[c for c in select_cols if c in g.columns])
df_preparado.createOrReplaceTempView("prep")

df_preparado.printSchema()
df_preparado.show(20, truncate=False)
```

```python
out_parquet = f"{BASE}/out_parquet/preparado.parquet"
(df_preparado
 .repartition(8)
 .write.mode("overwrite")
 .parquet(out_parquet))
print("Parquet escrito en:", out_parquet)
```

# Inciso 2

```python
from pyspark.sql import functions as F

BASE = "/opt/app/working_dir/"
parquet_path = f"{BASE}/out_parquet/preparado.parquet"   # <- el que guardaste

df_preparado = spark.read.parquet(parquet_path)
df_preparado.printSchema()
df_preparado.show(5, truncate=False)
```

```python
# mapa meses español -> número
mes_map = F.create_map(
    F.lit("enero"), F.lit(1), F.lit("febrero"), F.lit(2), F.lit("marzo"), F.lit(3),
    F.lit("abril"), F.lit(4), F.lit("mayo"), F.lit(5), F.lit("junio"), F.lit(6),
    F.lit("julio"), F.lit(7), F.lit("agosto"), F.lit(8), F.lit("septiembre"), F.lit(9),
    F.lit("octubre"), F.lit(10), F.lit("noviembre"), F.lit(11), F.lit("diciembre"), F.lit(12)
)

g = df_preparado

# mes_num
g = g.withColumn("mes_num", mes_map[F.lower(F.col("mes_ocu"))].cast("int"))

# hora_num (toma el primer grupo de dígitos como hora)
g = g.withColumn(
    "hora_num",
    F.regexp_extract(F.coalesce(F.col("hora_ocu"), F.lit("")), r"(\d{1,2})", 1).cast("int")
)

# severidad (2 fallecido, 1 lesionado, 0 ignorado/NaN)
g = g.withColumn(
    "sev_score",
    F.when(F.lower(F.col("fall_les")) == "fallecido", 2)
     .when(F.lower(F.col("fall_les")) == "lesionado", 1)
     .otherwise(0)
)

# tipos numéricos suaves
g = (g
     .withColumn("anio", F.col("anio").cast("int"))
     .withColumn("dia_ocu", F.col("dia_ocu").cast("int"))
     .withColumn("zona_ocu", F.col("zona_ocu").cast("int"))
     .withColumn("num_vehiculos_involucrados", F.col("num_vehiculos_involucrados").cast("int"))
)

g.select("anio","mes_ocu","mes_num","dia_ocu","hora_ocu","hora_num",
         "zona_ocu","num_vehiculos_involucrados","fall_les","sev_score").show(10, truncate=False)
```

```python
from pyspark.sql import functions as F

# Las mismas features que ibas a usar:
feature_cols = ["sev_score","num_vehiculos_involucrados","zona_ocu","anio","mes_num","dia_ocu","hora_num"]

# Reporte por columna: nulos, distintos y desviación estándar
stats = []
for c in feature_cols:
    stats.append(
        g.select(
            F.lit(c).alias("col"),
            F.count(F.when(F.col(c).isNull(), 1)).alias("n_null"),
            F.count("*").alias("n_total"),
            F.countDistinct(F.col(c)).alias("n_distinct"),
            F.stddev_samp(F.col(c).cast("double")).alias("stddev")
        )
    )

report = stats[0]
for s in stats[1:]:
    report = report.unionByName(s)

report.orderBy("col").show(truncate=False)
```

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation

# Selecciona automáticamente columnas con varianza > 0 y al menos 2 valores distintos
rep = report.collect()
valid_cols = [r["col"] for r in rep if r["n_distinct"] > 1 and r["stddev"] is not None and float(r["stddev"]) > 0.0]

print("Usando columnas para correlación:", valid_cols)

# Asegura que no queden nulos en esas columnas
g_num = g
for c in valid_cols:
    g_num = g_num.filter(F.col(c).isNotNull())

# Revisa cantidad de filas disponibles
print("Filas para correlación:", g_num.count())

# Si hay menos de 2 filas, Pearson no se puede calcular
if g_num.count() < 2 or len(valid_cols) < 2:
    raise ValueError("No hay suficientes filas o columnas válidas para calcular correlación.")

assembler = VectorAssembler(inputCols=valid_cols, outputCol="features", handleInvalid="skip")
g_vec = assembler.transform(g_num).select("features")

corr_mat = Correlation.corr(g_vec, "features", "pearson").head()[0]
```

```python
import numpy as np
import pandas as pd

corr_df = pd.DataFrame(np.array(corr_mat.toArray()), index=feature_cols, columns=feature_cols)
corr_df.round(3)
```

# Inciso 3

```python
from pyspark.sql import functions as F

# meses → número
mes_map = F.create_map(
    F.lit("enero"), F.lit(1),  F.lit("febrero"), F.lit(2),  F.lit("marzo"), F.lit(3),
    F.lit("abril"), F.lit(4),  F.lit("mayo"),    F.lit(5),  F.lit("junio"), F.lit(6),
    F.lit("julio"), F.lit(7),  F.lit("agosto"),  F.lit(8),  F.lit("septiembre"), F.lit(9),
    F.lit("octubre"),F.lit(10),F.lit("noviembre"),F.lit(11),F.lit("diciembre"),F.lit(12)
)

g = df_preparado

# mes_num
g = g.withColumn("mes_num", mes_map[F.lower(F.col("mes_ocu"))].cast("int"))

# hora_num (primer grupo de dígitos)
g = g.withColumn(
    "hora_num",
    F.regexp_extract(F.coalesce(F.col("hora_ocu"), F.lit("")), r"(\d{1,2})", 1).cast("int")
)

# indicadores de fallecido/lesionado (filas sin 'fall_les' → 0)
g = g.withColumn("fallecido_ind", F.when(F.lower(F.col("fall_les")) == "fallecido", 1).otherwise(0))
g = g.withColumn("lesionado_ind", F.when(F.lower(F.col("fall_les")) == "lesionado", 1).otherwise(0))

# castear otras columnas numéricas típicas
g = (g
     .withColumn("anio", F.col("anio").cast("int"))
     .withColumn("dia_ocu", F.col("dia_ocu").cast("int"))
     .withColumn("zona_ocu", F.col("zona_ocu").cast("int"))
     .withColumn("num_vehiculos_involucrados", F.col("num_vehiculos_involucrados").cast("int"))
)

g.select("anio","mes_num","dia_ocu","hora_num",
         "zona_ocu","num_vehiculos_involucrados",
         "fallecido_ind","lesionado_ind","tipo_eve","origen").show(10, truncate=False)
```

```python
# === PCA + extracción de PC1/PC2 vía RDD (Spark 3.1, sin UDF) ===
from pyspark.ml.feature import VectorAssembler, StandardScaler, PCA
from pyspark.ml.linalg import Vector
from pyspark.sql import functions as F

# 1) features
pca_features = [
    "hora_num",
    "num_vehiculos_involucrados",
    "lesionado_ind",
    "fallecido_ind",
    "zona_ocu",
    "anio",
    "mes_num",
    "dia_ocu",
]

# 2) filtra nulos
g_num = g
for c in pca_features:
    g_num = g_num.filter(F.col(c).isNotNull())

# 3) assembler → scaler → pca
va = VectorAssembler(inputCols=pca_features, outputCol="features", handleInvalid="skip")
vdf = va.transform(g_num)

scaler = StandardScaler(inputCol="features", outputCol="scaled", withStd=True, withMean=True)
sdf = scaler.fit(vdf).transform(vdf)

pca = PCA(k=2, inputCol="scaled", outputCol="pcaFeatures")
pca_model = pca.fit(sdf)
pca_df = pca_model.transform(sdf)
print("Explained variance (PC1, PC2):", pca_model.explainedVariance.toArray())

# 4) EXTRAER PC1/PC2 convirtiendo el vector a tupla de floats vía RDD
def vec_first_two(v: Vector):
    arr = v.toArray()
    return float(arr[0]), float(arr[1])

pca_plot = (pca_df
    .select("pcaFeatures","tipo_eve","origen")
    .rdd.map(lambda r: (*vec_first_two(r.pcaFeatures), r.tipo_eve, r.origen))
    .toDF(["PC1","PC2","tipo_eve","origen"])
)

pca_plot.show(10, truncate=False)
```

```python
import matplotlib.pyplot as plt
# --- pasar a pandas (con muestreo opcional para no reventar la RAM) ---
pdf = pca_plot.toPandas()
if len(pdf) > 50000:
    pdf = pdf.sample(n=50000, random_state=42)

# --- gráfico simple ---
ax = pdf.plot(kind="scatter", x="PC1", y="PC2", alpha=0.4, figsize=(6,5))
ax.set_title("PCA de Accidentes (PC1 vs PC2)")
ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
```

# Inciso 4

```python
from pyspark.sql import functions as F
from pyspark.ml.feature import VectorAssembler, StandardScaler

# === Features que usaremos en KMeans ===
km_features = [
    "hora_num",
    "num_vehiculos_involucrados",
    "lesionado_ind",
    "fallecido_ind",
    # puedes añadir si quieres:
    # "zona_ocu", "mes_num", "dia_ocu"
]

# Filtra filas con todas las features presentes
km_df = g
for c in km_features:
    km_df = km_df.filter(F.col(c).isNotNull())

print("Filas disponibles:", km_df.count())

# Conserva columnas útiles para interpretar luego (si existen)
interp_cols = [c for c in ["tipo_eve", "origen", "anio", "mes_ocu", "dia_semana"] if c in km_df.columns]

# Assemble -> Scale (estandarizado)
km_assembler = VectorAssembler(inputCols=km_features, outputCol="features", handleInvalid="skip")
km_vec = km_assembler.transform(km_df)

km_scaler = StandardScaler(inputCol="features", outputCol="scaled", withStd=True, withMean=True)
km_scaled = km_scaler.fit(km_vec).transform(km_vec).cache()

km_scaled.select("scaled").show(3, truncate=False)
```

```python
from pyspark.ml.clustering import KMeans
from pyspark.ml.evaluation import ClusteringEvaluator
import pandas as pd
import matplotlib.pyplot as plt

ks = [2, 3, 4, 5, 6]
rows = []

for k in ks:
    km = KMeans(k=k, seed=42, featuresCol="scaled", predictionCol="prediction")
    model = km.fit(km_scaled)
    preds = model.transform(km_scaled)

    wssse = model.summary.trainingCost  # Elbow
    sil = ClusteringEvaluator(featuresCol="scaled", predictionCol="prediction", metricName="silhouette").evaluate(preds)

    rows.append({"k": k, "WSSSE": wssse, "Silhouette": sil})

res = pd.DataFrame(rows)
print(res)

# Graficar codo (WSSSE) y Silhouette
fig, axes = plt.subplots(1,2, figsize=(10,4))
axes[0].plot(res["k"], res["WSSSE"], marker="o")
axes[0].set_title("Elbow (WSSSE)"); axes[0].set_xlabel("k"); axes[0].set_ylabel("WSSSE")

axes[1].plot(res["k"], res["Silhouette"], marker="o")
axes[1].set_title("Silhouette"); axes[1].set_xlabel("k"); axes[1].set_ylabel("Silhouette")
plt.tight_layout(); plt.show()

k_final = int(res.loc[res["Silhouette"].idxmax(), "k"])
print("k_final sugerido por Silhouette:", k_final)
```

```python
km_final = KMeans(k=int(k_final), seed=42, featuresCol="scaled", predictionCol="cluster")
model_final = km_final.fit(km_scaled)
asignados = model_final.transform(km_scaled)

conteo = asignados.groupBy("cluster").count().orderBy("cluster")
conteo.show(truncate=False)
```

```python
conteo = asignados.groupBy("cluster").count().orderBy("cluster")
conteo.show(truncate=False)
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Bandera de noche: 20–23 o 0–5
asignados_desc = asignados.withColumn(
    "is_noche", ((F.col("hora_num") >= 20) | (F.col("hora_num") <= 5)).cast("int")
)

carac = asignados_desc.groupBy("cluster").agg(
    F.count("*").alias("n"),
    F.avg("is_noche").alias("pct_noche"),
    F.avg("fallecido_ind").alias("pct_fallecido"),
    F.avg("lesionado_ind").alias("pct_lesionado"),
    F.avg("num_vehiculos_involucrados").alias("avg_vehiculos"),
    F.avg("hora_num").alias("avg_hora"),
    F.percentile_approx("hora_num", 0.5).alias("mediana_hora"),
)
carac.orderBy("cluster").show(truncate=False)

if "tipo_eve" in asignados_desc.columns:
    dist_tipo = asignados_desc.groupBy("cluster","tipo_eve").count()
    w = Window.partitionBy("cluster").orderBy(F.col("count").desc())
    top_tipo = (dist_tipo
                .withColumn("rn", F.row_number().over(w))
                .filter(F.col("rn")==1)
                .select("cluster","tipo_eve","count")
                .orderBy("cluster"))
    top_tipo.show(truncate=False)
```

```python
from pyspark.ml.feature import PCA
from pyspark.ml.feature import VectorAssembler, StandardScaler  # ya importados arriba
from pyspark.ml.linalg import Vector
import matplotlib.pyplot as plt

# Usamos las mismas features escaladas (columna 'scaled') para proyectar a 2D
pca = PCA(k=2, inputCol="scaled", outputCol="pcaFeatures")
pca_model = pca.fit(asignados)
pca_df = pca_model.transform(asignados)

# Extraer PC1/PC2 SIN UDF (RDD -> DataFrame) para Spark 3.1
def vec_first_two(v: Vector):
    arr = v.toArray()
    return float(arr[0]), float(arr[1])

pca_plot = (pca_df
    .select("pcaFeatures","cluster")
    .rdd.map(lambda r: (*vec_first_two(r.pcaFeatures), int(r.cluster)))
    .toDF(["PC1","PC2","cluster"])
)

# Muestra rápida y gráfico
pca_plot.show(5, truncate=False)

pdf = pca_plot.toPandas()
# muestrea si hay MUCHAS filas para que se vea mejor
if len(pdf) > 60000:
    pdf = pdf.sample(60000, random_state=42)

fig, ax = plt.subplots(figsize=(7,6))
for cl, sub in pdf.groupby("cluster"):
    ax.scatter(sub["PC1"], sub["PC2"], s=7, alpha=0.35, label=f"Cluster {cl}")
ax.set_title(f"PCA (PC1 vs PC2) por clúster (k={int(k_final)})")
ax.set_xlabel("PC1"); ax.set_ylabel("PC2")
ax.legend(title="cluster", fontsize=8)
plt.show()
```

```python
from pyspark.sql import functions as F

# Clave de evento (ajústala si tienes un ID más confiable)
key_cols = [c for c in ["anio","mes_num","dia_ocu","hora_num","depto_ocu","zona_ocu","tipo_eve"] if c in g.columns]

# Agregación por evento: conteos de fallecidos/lesionados
eventos = (
    g.groupBy(*key_cols)
     .agg(
        F.sum(F.col("fallecido_ind")).cast("int").alias("n_fallecidos"),
        F.sum(F.col("lesionado_ind")).cast("int").alias("n_lesionados"),
        # (opcional) conservar otras métricas del evento
        F.first("num_vehiculos_involucrados").alias("num_vehiculos_involucrados")
     )
)

# Regla de severidad
eventos = (
    eventos.withColumn(
        "severidad",
        F.when( (F.col("n_fallecidos")==0) & (F.col("n_lesionados") < 2), "Leve")
         .when( (F.col("n_fallecidos")==1) | ((F.col("n_lesionados")>=2) & (F.col("n_lesionados")<=5)), "Moderado")
         .otherwise("Grave")
    )
)

eventos.select(*key_cols, "n_fallecidos","n_lesionados","severidad").show(10, truncate=False)
```

```python
from pyspark.sql import functions as F
import matplotlib.pyplot as plt

# Conteo y % por categoría
dist = eventos.groupBy("severidad").count()

total = eventos.count()
dist = dist.withColumn("porcentaje", (F.col("count")/F.lit(total)*100))

# Orden lógico de las categorías
orden = F.when(F.col("severidad")=="Leve", 1).when(F.col("severidad")=="Moderado", 2).otherwise(3)
dist = dist.orderBy(orden)

dist.show(truncate=False)

# (Opcional) gráfico simple
pdf_dist = dist.toPandas()
ax = pdf_dist.plot(kind="bar", x="severidad", y="count", legend=False, rot=0)
ax.set_title("Distribución de severidad de accidentes")
ax.set_xlabel("Severidad")
ax.set_ylabel("Número de accidentes")
for i, v in enumerate(pdf_dist["porcentaje"].round(1)):
    ax.text(i-0.15, pdf_dist.loc[i,"count"]*1.02, f"{v}%", fontsize=9)
plt.show()
```

# Inciso 7

```python
# ============================================
# Inciso 7 – Pipeline de Clasificación (Spark) ROBUSTO + maxBins dinámico
# Etiqueta: 'origen' (ajústala si deseas otra)
# ============================================
from pyspark.sql import functions as F
from pyspark.sql.functions import col, isnan, sum as Fsum
from pyspark.ml import Pipeline
from pyspark.ml.feature import (
    StringIndexer, VectorAssembler, StandardScaler,
    Imputer, VectorIndexer
)
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
from pyspark.mllib.evaluation import MulticlassMetrics

print("Columnas del DataFrame:", df_preparado.columns)

# ---------------------------
# 1) Definición de columnas
# ---------------------------
LABEL_COL = "origen"  # Cambia a "tipo_eve" si prefieres
if LABEL_COL not in df_preparado.columns:
    raise ValueError(f"No existe la columna de etiqueta '{LABEL_COL}'.")

categorical_cols = [c for c in ["tipo_eve", "depto_ocu", "zona_ocu", "dia_semana"] if c in df_preparado.columns]
numeric_cols_base = [c for c in ["hora_ocu", "num_vehiculos_involucrados", "fall_les"] if c in df_preparado.columns]

# Cast numéricas
df_cast = df_preparado
for c in numeric_cols_base:
    df_cast = df_cast.withColumn(c, F.col(c).cast("double"))

# Filtrar filas sin etiqueta
df_clean = df_cast.dropna(subset=[LABEL_COL])

# ---------------------------
# 2) Split + detección de NaN en TRAIN
# ---------------------------
train_df, test_df = df_clean.randomSplit([0.8, 0.2], seed=42)

# Numéricas con al menos un valor válido (no NaN/Null) en TRAIN
valid_counts_row = train_df.select([
    Fsum((~(isnan(col(c)) | col(c).isNull())).cast("int")).alias(c)
    for c in numeric_cols_base
]).collect()[0].asDict()
available_num = [c for c in numeric_cols_base if valid_counts_row.get(c, 0) > 0]

dropped_all_missing_train = list(set(numeric_cols_base) - set(available_num))
if dropped_all_missing_train:
    print("⚠️ Columnas numéricas 100% NaN/Null en TRAIN, excluidas del Imputer:", dropped_all_missing_train)

use_numeric = len(available_num) > 0

# ---------------------------
# 3) Cardinalidad categórica (para maxBins y VectorIndexer)
# ---------------------------
# Contamos niveles por columna categórica en TRAIN (sin nulos)
cat_card = {}
for c in categorical_cols:
    n_dist = train_df.select(c).where(col(c).isNotNull()).distinct().count()
    cat_card[c] = int(n_dist)

max_cat_levels = max(cat_card.values()) if cat_card else 0
print("Cardinalidades categóricas TRAIN:", cat_card)

# Reglas: maxBins debe ser >= niveles máximos
# Usamos un piso 32 y un tope razonable para no explotar memoria
rf_maxBins = max(32, max_cat_levels)
# Si ves OOM, puedes bajar profundidad/árboles; si sale error de bins, sube este valor:
# rf_maxBins = max(rf_maxBins, 64)

# El VectorIndexer también debe permitir al menos esa cardinalidad
vector_indexer_maxCats = max(32, max_cat_levels)

# ---------------------------
# 4) Stages del pipeline
# ---------------------------
label_indexer = StringIndexer(inputCol=LABEL_COL, outputCol="label_idx", handleInvalid="skip")

indexers = [
    StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="skip")
    for c in categorical_cols
]

stages_previas = [label_indexer] + indexers
final_feature_cols = []

if use_numeric:
    imputer = Imputer(
        inputCols=available_num,
        outputCols=[f"{c}_imp" for c in available_num]
    )
    num_assembler = VectorAssembler(
        inputCols=[f"{c}_imp" for c in available_num],
        outputCol="num_vec"
    )
    scaler = StandardScaler(inputCol="num_vec", outputCol="num_scaled", withMean=True, withStd=True)

    stages_previas += [imputer, num_assembler, scaler]
    final_feature_cols.append("num_scaled")

# Añadir categóricas indexadas
final_feature_cols += [f"{c}_idx" for c in categorical_cols]
if not final_feature_cols:
    raise ValueError("No hay columnas válidas para 'features'.")

features_assembler = VectorAssembler(inputCols=final_feature_cols, outputCol="features")
stages_previas += [features_assembler]

# VectorIndexer: marca dentro del vector cuáles son categóricas
vector_indexer = VectorIndexer(
    inputCol="features",
    outputCol="features_indexed",
    maxCategories=vector_indexer_maxCats
)
stages_previas += [vector_indexer]

# ---------------------------
# 5) RandomForest ajustado (evita OOM) + maxBins dinámico
# ---------------------------
rf_tuned = RandomForestClassifier(
    labelCol="label_idx",
    featuresCol="features_indexed",
    numTrees=50,            # baja complejidad
    maxDepth=9,
    maxBins=rf_maxBins,     # <- clave para tu error
    subsamplingRate=0.7,
    minInstancesPerNode=20,
    seed=42
)

# ---------------------------
# 6) Pipeline y entrenamiento
# ---------------------------
stages = stages_previas + [rf_tuned]
pipeline = Pipeline(stages=stages)

# Cache para evitar recomputaciones pesadas
train_df = train_df.cache(); _ = train_df.count()

print(f"\nEntrenando RandomForest... (maxBins={rf_maxBins}, maxCategories(VectorIndexer)={vector_indexer_maxCats})")
model = pipeline.fit(train_df)

# ---------------------------
# 7) Predicciones y métricas
# ---------------------------
pred = model.transform(test_df).cache()
pred.select("features_indexed", "label_idx", "prediction").show(10, truncate=False)

evaluator = MulticlassClassificationEvaluator(labelCol="label_idx", predictionCol="prediction")
eval_acc  = evaluator.setMetricName("accuracy").evaluate(pred)
eval_f1   = evaluator.setMetricName("f1").evaluate(pred)
eval_prec = evaluator.setMetricName("weightedPrecision").evaluate(pred)
eval_rec  = evaluator.setMetricName("weightedRecall").evaluate(pred)

print("=== Métricas globales (ponderadas) ===")
print(f"Accuracy : {eval_acc:.4f}")
print(f"Precision: {eval_prec:.4f}")
print(f"Recall   : {eval_rec:.4f}")
print(f"F1-score : {eval_f1:.4f}")

# (Opcional) por clase
pred_rdd = pred.select(F.col("prediction").cast("double"), F.col("label_idx").cast("double")).rdd.map(tuple)
metrics = MulticlassMetrics(pred_rdd)
print("\n=== Métricas por clase (precision, recall, f1) ===")
for lbl in sorted([float(r['label_idx']) for r in pred.select("label_idx").distinct().collect()]):
    print(f"Clase {lbl:.0f} -> P:{metrics.precision(lbl):.4f} | R:{metrics.recall(lbl):.4f} | F1:{metrics.fMeasure(lbl):.4f}")

# Importancias
print("\n=== Importancia de características (top 15) ===")
rf_model = model.stages[-1]
importances = rf_model.featureImportances.toArray()

feature_names = []
if use_numeric:
    feature_names += [f"num_scaled[{i}] ({col})" for i, col in enumerate(available_num)]
feature_names += [f"{c}_idx" for c in categorical_cols]

pairs = list(zip(feature_names, importances[:len(feature_names)]))
for name, imp in sorted(pairs, key=lambda x: x[1], reverse=True)[:15]:
    print(f"{name:30s}  {imp:.6f}")


```

# Inciso 8

```python
# ============================================
# Inciso 8 – Comparar modelos (RF, LR, DT) ROBUSTO
# Mismo preprocesamiento del inciso 7: Imputer seguro, VectorIndexer, maxBins dinámico
# ============================================
from pyspark.sql import functions as F
from pyspark.sql.functions import col, isnan, sum as Fsum
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, VectorAssembler, StandardScaler, Imputer, VectorIndexer
from pyspark.ml.classification import RandomForestClassifier, LogisticRegression, DecisionTreeClassifier
from pyspark.ml.evaluation import MulticlassClassificationEvaluator

print("Columnas df_preparado:", df_preparado.columns)

# ---------- 1) Columnas ----------
LABEL_COL = "origen"  # cambia a "tipo_eve" si prefieres
if LABEL_COL not in df_preparado.columns:
    raise ValueError(f"No existe la columna de etiqueta '{LABEL_COL}' en df_preparado.")

categorical_cols = [c for c in ["tipo_eve", "depto_ocu", "zona_ocu", "dia_semana"] if c in df_preparado.columns]
numeric_cols_base = [c for c in ["hora_ocu", "num_vehiculos_involucrados", "fall_les"] if c in df_preparado.columns]

# Cast numéricas
df_cast = df_preparado
for c in numeric_cols_base:
    df_cast = df_cast.withColumn(c, F.col(c).cast("double"))

# Quitar filas sin etiqueta
df_clean = df_cast.dropna(subset=[LABEL_COL])

# ---------- 2) Split + columnas numéricas válidas en TRAIN ----------
train_df, test_df = df_clean.randomSplit([0.8, 0.2], seed=42)

valid_counts_row = train_df.select([
    Fsum((~(isnan(col(c)) | col(c).isNull())).cast("int")).alias(c)
    for c in numeric_cols_base
]).collect()[0].asDict()
available_num = [c for c in numeric_cols_base if valid_counts_row.get(c, 0) > 0]

if set(numeric_cols_base) - set(available_num):
    print("⚠️ Columnas 100% NaN/Null en TRAIN, excluidas del Imputer:",
          list(set(numeric_cols_base) - set(available_num)))

use_numeric = len(available_num) > 0

# ---------- 3) Cardinalidad categórica en TRAIN (para maxBins / VectorIndexer) ----------
cat_card = {c: train_df.select(c).where(col(c).isNotNull()).distinct().count() for c in categorical_cols}
max_cat_levels = max(cat_card.values()) if cat_card else 0
print("Cardinalidades categóricas TRAIN:", cat_card)

# maxBins >= #niveles categóricos
rf_maxBins = max(32, max_cat_levels)
dt_maxBins = rf_maxBins
vi_maxCats = max(32, max_cat_levels)

# ---------- 4) Etapas previas compartidas ----------
label_indexer = StringIndexer(inputCol=LABEL_COL, outputCol="label_idx", handleInvalid="skip")
indexers = [StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="skip") for c in categorical_cols]

stages_previas = [label_indexer] + indexers
final_feature_cols = []

if use_numeric:
    imputer = Imputer(
        inputCols=available_num,
        outputCols=[f"{c}_imp" for c in available_num]
    )
    num_assembler = VectorAssembler(inputCols=[f"{c}_imp" for c in available_num], outputCol="num_vec")
    scaler = StandardScaler(inputCol="num_vec", outputCol="num_scaled", withMean=True, withStd=True)
    stages_previas += [imputer, num_assembler, scaler]
    final_feature_cols.append("num_scaled")

final_feature_cols += [f"{c}_idx" for c in categorical_cols]
if not final_feature_cols:
    raise ValueError("No hay columnas válidas para 'features'.")

features_assembler = VectorAssembler(inputCols=final_feature_cols, outputCol="features")
stages_previas += [features_assembler]

vector_indexer = VectorIndexer(
    inputCol="features",
    outputCol="features_indexed",
    maxCategories=vi_maxCats
)
stages_previas += [vector_indexer]

# ---------- 5) Modelos “ligeros” (evitar OOM) ----------
rf_modelo = RandomForestClassifier(
    labelCol="label_idx",
    featuresCol="features_indexed",
    numTrees=20,           # ↓ árboles
    maxDepth=7,           # ↓ profundidad
    maxBins=rf_maxBins,   # >= categorías
    subsamplingRate=0.6,  # muestreo por árbol
    minInstancesPerNode=50,
    seed=42
)

dt_modelo = DecisionTreeClassifier(
    labelCol="label_idx",
    featuresCol="features_indexed",
    maxDepth=7,
    maxBins=dt_maxBins,
    minInstancesPerNode=50,
    seed=42
)

lr_modelo = LogisticRegression(
    labelCol="label_idx",
    featuresCol="features_indexed",
    maxIter=100,
    regParam=0.02,
    elasticNetParam=0.8
)

# ---------- 6) Evaluación común ----------
def evaluar_modelo(modelo, nombre, train_df, test_df, stages_previas):
    pipe = Pipeline(stages=stages_previas + [modelo])

    # Cachear train para evitar recomputaciones pesadas
    train_df_cache = train_df.cache()
    _ = train_df_cache.count()

    mdl = pipe.fit(train_df_cache)
    pred = mdl.transform(test_df).cache()

    ev = MulticlassClassificationEvaluator(labelCol="label_idx", predictionCol="prediction")
    return {
        "Modelo": nombre,
        "Accuracy": ev.setMetricName("accuracy").evaluate(pred),
        "Precision": ev.setMetricName("weightedPrecision").evaluate(pred),
        "Recall": ev.setMetricName("weightedRecall").evaluate(pred),
        "F1-score": ev.setMetricName("f1").evaluate(pred),
    }

# ---------- 7) Entrenar y comparar ----------
resultados = []
for mdl, nombre in [
    (rf_modelo, "Random Forest"),
    (lr_modelo, "Logistic Regression"),
    (dt_modelo, "Decision Tree"),
]:
    print(f"\nEntrenando y evaluando: {nombre} (maxBins={rf_maxBins}, maxCats={vi_maxCats})")
    resultados.append(evaluar_modelo(mdl, nombre, train_df, test_df, stages_previas))

# ---------- 8) Tabla y discusión (formateo seguro) ----------
try:
    import pandas as pd
    tabla = pd.DataFrame(resultados)
    # Formatear SOLO columnas numéricas
    num_cols = ["Accuracy", "Precision", "Recall", "F1-score"]
    for c in num_cols:
        if c in tabla.columns:
            tabla[c] = tabla[c].astype(float).round(4)
    print("\n=== Comparación de modelos ===")
    display(tabla)  # sin Styler.format, para evitar el ValueError
except Exception:
    print("\n=== Comparación de modelos ===")
    for r in resultados:
        print({k: (round(v,4) if isinstance(v,float) else v) for k,v in r.items()})

best = max(resultados, key=lambda r: r["F1-score"])
print(f"\nMejor F1-score: {best['Modelo']} ({best['F1-score']:.4f})")
if best["Modelo"] == "Random Forest":
    print(" Suele ganar con relaciones no lineales y varias categóricas; el ensamble reduce varianza.")
elif best["Modelo"] == "Logistic Regression":
    print("Favorecido si la separación es aproximadamente lineal y las variables están bien escaladas.")
else:
    print("Interpretabilidad alta; con profundidad limitada puede rendir menos que RF.")

```

```python
from pyspark.sql import functions as F
from pyspark.ml.feature import StringIndexer, Imputer, VectorAssembler, StandardScaler
from pyspark.ml.regression import LinearRegression
from pyspark.ml.evaluation import RegressionEvaluator
from pyspark.ml import Pipeline

# === Configuración de columnas (según tu schema real) ===
label = "n_fallecidos"
num_cols = ["hora_num", "num_vehiculos_involucrados", "n_lesionados"]
cat_cols = ["tipo_eve", "zona_ocu"]  # serán indexadas en el pipeline

# --- Helpers: casteos seguros ---
def cast_to_double(df, cols):
    for c in cols:
        if c in df.columns:
            df = df.withColumn(c, F.col(c).cast("double"))
    return df

def cast_to_string(df, cols):
    for c in cols:
        if c in df.columns:
            df = df.withColumn(c, F.col(c).cast("string"))
    return df

# 1) Asegura tipos correctos (numéricas -> double, categóricas -> string, label -> double)
train2 = cast_to_double(train,  num_cols + [label])
test2  = cast_to_double(test,   num_cols + [label])

train2 = cast_to_string(train2, cat_cols)
test2  = cast_to_string(test2,  cat_cols)

# 2) Quita nulos en la etiqueta (imprescindible)
train2 = train2.na.drop(subset=[label])
test2  = test2.na.drop(subset=[label])

# 3) Stages del pipeline
# 3a) Indexers para categóricas (si el col existe)
indexers = []
cat_idx_cols = []
for c in cat_cols:
    if c in train2.columns:
        idx = f"{c}_idx"
        cat_idx_cols.append(idx)
        indexers.append(
            StringIndexer(
                inputCol=c,
                outputCol=idx,
                handleInvalid="keep"
            )
        )

# 3b) Imputer para numéricas (mediana, robusto)
imp_out = [f"{c}_imp" for c in num_cols]
imputer = Imputer(inputCols=num_cols, outputCols=imp_out, strategy="median")

# 3c) VectorAssembler (usa idx + numéricas imputadas)
assembler = VectorAssembler(
    inputCols=cat_idx_cols + imp_out,
    outputCol="features",
    handleInvalid="keep"   # <- clave para evitar fallos por NaN residuales
)

# 3d) Escalado (opcional pero recomendado)
scaler = StandardScaler(inputCol="features", outputCol="scaledFeatures", withMean=True, withStd=True)

# 3e) Modelo
lr = LinearRegression(featuresCol="scaledFeatures", labelCol=label)

# 4) Pipeline completo
pipeline_lr = Pipeline(stages=indexers + [imputer, assembler, scaler, lr])

# 5) Entrenar y predecir
model_lr = pipeline_lr.fit(train2)
pred_lr = model_lr.transform(test2)

# 6) Métricas
e_rmse = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="rmse")
e_mae  = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="mae")
e_r2   = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="r2")

print(f"RMSE: {e_rmse.evaluate(pred_lr):.3f}")
print(f"MAE : {e_mae.evaluate(pred_lr):.3f}")
print(f"R²  : {e_r2.evaluate(pred_lr):.3f}")

pred_lr.select(
    "tipo_eve", "zona_ocu", "hora_num", "num_vehiculos_involucrados",
    "n_lesionados", "n_fallecidos", "prediction"
).show(10, truncate=False)
```

```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder, Imputer, VectorAssembler, StandardScaler
from pyspark.ml.regression import LinearRegression
from pyspark.ml import Pipeline
from pyspark.sql import functions as F

label = "n_fallecidos"
num_cols = ["hora_num", "num_vehiculos_involucrados", "n_lesionados"]
cat_cols = ["tipo_eve", "zona_ocu"]

# Tipos correctos
def cast_to_double(df, cols):
    for c in cols:
        if c in df.columns:
            df = df.withColumn(c, F.col(c).cast("double"))
    return df
def cast_to_string(df, cols):
    for c in cols:
        if c in df.columns:
            df = df.withColumn(c, F.col(c).cast("string"))
    return df

train3 = cast_to_double(train,  num_cols + [label])
test3  = cast_to_double(test,   num_cols + [label])
train3 = cast_to_string(train3, cat_cols)
test3  = cast_to_string(test3,  cat_cols)

train3 = train3.na.drop(subset=[label])
test3  = test3.na.drop(subset=[label])

# Index + OneHot
indexers = [StringIndexer(inputCol=c, outputCol=f"{c}_idx", handleInvalid="keep") for c in cat_cols]
encoders = [OneHotEncoder(inputCols=[f"{c}_idx"], outputCols=[f"{c}_oh"], handleInvalid="keep") for c in cat_cols]

# Imputer numérico
imp_out = [f"{c}_imp" for c in num_cols]
imputer = Imputer(inputCols=num_cols, outputCols=imp_out, strategy="median")

# Ensamblado
assembler = VectorAssembler(
    inputCols=[f"{c}_oh" for c in cat_cols] + imp_out,
    outputCol="features",
    handleInvalid="keep"
)

scaler = StandardScaler(inputCol="features", outputCol="scaledFeatures", withMean=True, withStd=True)

# Regresión lineal regularizada
lr = LinearRegression(
    featuresCol="scaledFeatures",
    labelCol=label,
    regParam=0.1,            # > 0 para estabilizar
    elasticNetParam=0.5,     # mezcla L1/L2
    fitIntercept=True
)

pipe = Pipeline(stages=indexers + encoders + [imputer, assembler, scaler, lr])

model_lr2 = pipe.fit(train3)
pred_lr2 = model_lr2.transform(test3)

from pyspark.ml.evaluation import RegressionEvaluator
e_rmse = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="rmse")
e_mae  = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="mae")
e_r2   = RegressionEvaluator(labelCol=label, predictionCol="prediction", metricName="r2")

print(f"RMSE: {e_rmse.evaluate(pred_lr2):.3f}")
print(f"MAE : {e_mae.evaluate(pred_lr2):.3f}")
print(f"R²  : {e_r2.evaluate(pred_lr2):.3f}")
```

```python
model_path = "/opt/app/working_dir/modelo_fallecidos.model"
model_lr2.write().overwrite().save(model_path)
print(f"✅ Modelo guardado en: {model_path}")
```

```python
# 10) Recargar el modelo guardado y probarlo sobre un subconjunto nuevo (año 2023)

from pyspark.ml.pipeline import PipelineModel
from pyspark.sql import functions as F
from pyspark.ml.evaluation import RegressionEvaluator

# === 1) Cargar el modelo guardado ===
model_path = "/opt/app/working_dir/modelo_fallecidos.model"
modelo_cargado = PipelineModel.load(model_path)

# === 2) Usa el DataFrame base que tengas ===
# Cambia aquí por el nombre que realmente usaste (train, df_reg, df_final, etc.)
# 👇👇👇👇👇
base_df = train.union(test)  # o reemplaza por df_reg / df_final
# 👆👆👆👆👆

# Filtrar subconjunto del año 2023 (o el más reciente si no hay)
if "anio" not in base_df.columns:
    raise ValueError("No existe la columna 'anio' en el DataFrame. Verifica el nombre (quizás 'año').")

subset = base_df.filter(F.col("anio") == 2023)
if subset.rdd.isEmpty():
    max_year = base_df.agg(F.max("anio")).first()[0]
    print(f"[INFO] No hay filas para 2023. Usando el año más reciente: {max_year}")
    subset = base_df.filter(F.col("anio") == max_year)

# Limpieza: quitar nulos en la etiqueta
subset = subset.na.drop(subset=["n_fallecidos"])

# === 3) Predecir con el modelo cargado ===
pred_nuevo = modelo_cargado.transform(subset)

# === 4) Mostrar resultados reales vs predichos ===
pred_nuevo.select(
    "anio", "tipo_eve", "zona_ocu",
    "hora_num", "num_vehiculos_involucrados", "n_lesionados",
    "n_fallecidos", F.col("prediction").alias("fallecidos_pred")
).show(20, truncate=False)

# === 5) Métricas ===
e_rmse = RegressionEvaluator(labelCol="n_fallecidos", predictionCol="prediction", metricName="rmse")
e_mae  = RegressionEvaluator(labelCol="n_fallecidos", predictionCol="prediction", metricName="mae")
e_r2   = RegressionEvaluator(labelCol="n_fallecidos", predictionCol="prediction", metricName="r2")

print(f"RMSE (subconjunto): {e_rmse.evaluate(pred_nuevo):.3f}")
print(f"MAE  (subconjunto): {e_mae.evaluate(pred_nuevo):.3f}")
print(f"R²   (subconjunto): {e_r2.evaluate(pred_nuevo):.3f}")
```

```python
# 11) Graficar predicho vs real y analizar comportamiento

import matplotlib.pyplot as plt
import pandas as pd

# Convertir las columnas necesarias a pandas
pdf_pred = pred_nuevo.select("n_fallecidos", "prediction").toPandas()

# --- Crear la gráfica ---
plt.figure(figsize=(6,6))
plt.scatter(pdf_pred["n_fallecidos"], pdf_pred["prediction"], alpha=0.6, edgecolors='k')
plt.plot([0, pdf_pred["n_fallecidos"].max()], [0, pdf_pred["n_fallecidos"].max()],
         color="red", linestyle="--", label="Ideal (y=x)")
plt.xlabel("Fallecidos reales")
plt.ylabel("Fallecidos predichos")
plt.title("Comparación entre valores reales y predichos (Regresión lineal)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# En la gráfica de comparación entre valores reales y predichos se observa que la mayoría de los puntos se agrupan muy cerca del eje horizontal, mientras que la línea roja (y = x) representa la predicción ideal.
# Esto indica que el modelo tiende a predecir valores cercanos a cero incluso cuando el número real de fallecidos es mayor.
# En otras palabras, el modelo subestima los valores altos de fallecidos, mostrando dificultad para captar los accidentes más severos.
# Este comportamiento se debe principalmente al desequilibrio de los datos, ya que la gran mayoría de los accidentes presentan cero o un fallecido, lo que hace que la regresión lineal minimice el error concentrándose en los casos más frecuentes.
# A pesar de ello, el modelo mantiene errores absolutos bajos (MAE ≈ 0.3) y cumple su objetivo de estimar de manera general la severidad de los accidentes.
```
